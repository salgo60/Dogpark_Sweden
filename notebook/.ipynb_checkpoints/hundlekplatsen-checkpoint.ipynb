```python
import asyncio
from playwright.async_api import async_playwright

BASE_URL = "https://www.hundlekplatsen.se/cities"

async def sniff_network(city):
    print(f"\n### Kollar {city} ###")
    async with async_playwright() as p:
        browser = await p.firefox.launch(headless=True)
        page = await browser.new_page()

        async def log_response(response):
            if response.request.resource_type == "xhr" or response.request.resource_type == "fetch":
                print(response.url)

        page.on("response", log_response)
        await page.goto(f"{BASE_URL}/{city}")
        await page.wait_for_load_state("networkidle")
        await asyncio.sleep(4)
        await browser.close()

await sniff_network("Stockholm")
```

```python
import asyncio
import json
import re
from playwright.async_api import async_playwright

CITY = "Stockholm"
BASE_URL = "https://www.hundlekplatsen.se"

async def get_city_places(page, city):
    """Hämtar alla hundparkslänkar för en stad"""
    url = f"{BASE_URL}/cities/{city}"
    print(f"📍 Hämtar stadssida: {url}")
    await page.goto(url)
    await page.wait_for_load_state("networkidle")
    await asyncio.sleep(4)

    links = await page.query_selector_all("a[href*='/dogpark/']")
    places = []
    for link in links:
        href = await link.get_attribute("href")
        name = (await link.inner_text()).strip()
        if href and href.startswith("/dogpark/"):
            places.append({
                "city": city,
                "name": name,
                "url": f"{BASE_URL}{href}"
            })
    print(f"✅ Hittade {len(places)} länkar i {city}")
    return places


async def get_place_details(page, place):
    """Besöker en plats och hämtar metadata"""
    await page.goto(place["url"])
    await page.wait_for_load_state("networkidle")
    await asyncio.sleep(2)

    # --- Beskrivning ---
    desc_el = await page.query_selector("p, div.text-sm, div.text-base")
    desc = (await desc_el.inner_text()).strip() if desc_el else ""

    # --- Bild ---
    img_el = await page.query_selector("img")
    img = await img_el.get_attribute("src") if img_el else None
    if img and img.startswith("/"):
        img = BASE_URL + img

    # --- Koordinater ---
    lat, lon = None, None
    iframe = await page.query_selector("iframe[src*='google']")
    if iframe:
        src = await iframe.get_attribute("src")
        m1 = re.search(r"!3d([0-9.\-]+)!4d([0-9.\-]+)", src or "")
        m2 = re.search(r"!2d([0-9.\-]+)!3d([0-9.\-]+)", src or "")
        if m1:
            lat, lon = m1.group(1), m1.group(2)
        elif m2:
            lon, lat = m2.group(1), m2.group(2)

    # --- Vägbeskrivning (fallback) ---
    if not lat or not lon:
        map_links = await page.query_selector_all("a[href*='google.com/maps']")
        for a in map_links:
            href = await a.get_attribute("href")
            if not href:
                continue
            m3 = re.search(r"@([0-9.\-]+),([0-9.\-]+)", href)
            if m3:
                lat, lon = m3.group(1), m3.group(2)
                break
            m4 = re.search(r"!3d([0-9.\-]+)!4d([0-9.\-]+)", href)
            m5 = re.search(r"!2d([0-9.\-]+)!3d([0-9.\-]+)", href)
            if m4:
                lat, lon = m4.group(1), m4.group(2)
                break
            elif m5:
                lon, lat = m5.group(1), m5.group(2)
                break

    # --- Metadata / features ---
    features = []
    feature_els = await page.query_selector_all("li, div")
    for el in feature_els:
        text = (await el.inner_text()).strip().lower()
        if any(keyword in text for keyword in [
            "bana", "agility", "vatten", "toalett", "belysning",
            "inhägnad", "gräs", "grus", "skugga", "öppet", "parkering"
        ]):
            features.append(text)

    # --- Sammanställ ---
    place.update({
        "description": desc,
        "image": img,
        "latitude": lat,
        "longitude": lon,
        "features": list(set(features))  # ta bort dubbletter
    })
    print(f"  ↳ {place['name']} ({lat}, {lon})")
    return place


async def main():
    async with async_playwright() as p:
        browser = await p.firefox.launch(headless=True)
        page = await browser.new_page()

        # Steg 1: Hämta alla platser i staden
        places = await get_city_places(page, CITY)

        # Steg 2: Gå igenom varje plats och hämta detaljer
        detailed = []
        for i, place in enumerate(places, start=1):
            try:
                print(f"\n[{i}/{len(places)}] Hämtar detaljer för {place['name']}")
                detailed.append(await get_place_details(page, place))
            except Exception as e:
                print(f"⚠️ Fel vid {place['url']}: {e}")

        await browser.close()

    # Steg 3: Spara resultat
    with open(f"hundlekplatser_{CITY.lower()}.json", "w", encoding="utf-8") as f:
        json.dump(detailed, f, ensure_ascii=False, indent=2)

    print(f"\n🎉 Sparade {len(detailed)} platser i hundlekplatser_{CITY.lower()}.json")


# --- Kör allt ---
await main()
    
```
