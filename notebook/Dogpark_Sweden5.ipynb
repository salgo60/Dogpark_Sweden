```python
import pandas as pd

def safe_popup_row(row):
    """
    Säkerställer att en rad (dict eller Pandas Series) har alla fält som build_popup() behöver,
    och att alla är rätt datatyp (str, dict, float osv.)
    """
    # Om rad är en Pandas Series → gör till dict
    if not isinstance(row, dict):
        row = row.to_dict()

    required_fields = [
        "lat", "lon", "name", "description", "website", "url", "image",
        "wikidata", "operator:wikidata", "tags", "type", "id"
    ]
    for field in required_fields:
        row.setdefault(field, "")

    # Rensa textfält
    for field in [
        "name", "description", "website", "url", "image",
        "wikidata", "operator:wikidata", "type", "id"
    ]:
        val = row.get(field, "")
        if pd.isna(val):
            val = ""
        if not isinstance(val, str):
            val = str(val)
        row[field] = val.strip()

    # Hantera tags: se till att det är en dict
    tags_val = row.get("tags", {})
    if not isinstance(tags_val, dict):
        if isinstance(tags_val, str):
            try:
                import json
                tags_val = json.loads(tags_val)
            except Exception:
                tags_val = {}
        else:
            tags_val = {}
    row["tags"] = tags_val

    # Säkerställ att lat/lon är numeriska
    try:
        row["lat"] = float(row.get("lat", 0))
        row["lon"] = float(row.get("lon", 0))
    except Exception:
        row["lat"] = 0.0
        row["lon"] = 0.0

    return row

from urllib.parse import urlparse
import pandas as pd

from urllib.parse import urlparse
import pandas as pd

def build_popup(row, source=None):
    """
    Bygger en standardiserad popup med stöd för källinformation och förenklad logik.
    'source' styr både layout, etiketter och domänigenkänning.
    """

    # ------------- Basdata -------------
    lat = row.get("lat", None)
    lon = row.get("lon", None)
    name = row.get("name", "Okänd hundrastgård")
    desc = row.get("description", "")
    website = (row.get("website") or "").strip()
    url = (row.get("url") or "").strip()  # ev. extern källa
    image = row.get("image", "")
    wikidata = row.get("wikidata", "")
    operator_wd = row.get("operator:wikidata", "")
    osm_type = row.get("type", "")
    osm_id = row.get("id", "")
    tags = row.get("tags", {})

    # ------------- Fördefinierade källor -------------
    sources = {
        "osm": {
            "id": "osm",
            "name": "OpenStreetMap",
            "type": "Crowdsourcad databas",
            "copyright": "© OpenStreetMap contributors",
            "copyright_link": "https://www.openstreetmap.org/copyright",
            "link": "https://www.openstreetmap.org/",
            "license": "Open Database License (ODbL) 1.0",
            "license_link": "https://opendatacommons.org/licenses/odbl/1-0/",
            "icon": "🗺️"
        },     
        "hundlistan": {
            "name": "Hundlistan.se",
            "type": "crowdsourced",
            "domain": "hundlistan.se",
            "copyright": "okänd",
            "link": "",
            "icon": "🐶"
        },
          "hundlekplatsen": {
            "id": "Hundlekplatsen",
            "name": "hundlekplatsen.se",
            "type": "crowdsourced",
            "copyright": "okänd",
            "copyright_link": "",
            "link": "",
            "icon": "🐕" 
        },
        "stockholm": {
            "name": "Stockholms stad",
            "type": "kommun",
            "domain": "stockholm.se",
            "copyright": "CC-0",
            "link": "https://creativecommons.org/publicdomain/zero/1.0/",
            "icon": "🟠"
        },
        "sodertalje": {
            "name": "Södertälje kommun",
            "type": "kommun",
            "domain": "sodertalje.se",
            "copyright": "CC-0",
            "link": "https://creativecommons.org/publicdomain/zero/1.0/",
            "icon": "🐾"
        },
        "boras": {
            "name": "Borås kommun",
            "type": "kommun",
            "domain": "boras.se",
            "copyright": "CC-0",
            "link": "https://creativecommons.org/publicdomain/zero/1.0/",
            "icon": "🐾"
        },
        "gavle": {
            "name": "Gävle kommun",
            "type": "kommun",
            "domain": "gavle.se",
            "copyright": "CC-0",
            "link": "https://creativecommons.org/publicdomain/zero/1.0/",
            "icon": "🐾"
        },
        "botkyrka": {
            "name": "Botkyrka kommun",
            "type": "kommun",
            "domain": "botkyrka.se",
            "copyright": "CC-0",
            "link": "https://creativecommons.org/publicdomain/zero/1.0/",
            "icon": "🐕"
        },
        "commons": {
            "name": "Wikimedia Commons",
            "type": "commons",
            "domain": "commons.wikimedia.org",
            "copyright": "Varierar per bild (ofta CC-0)",
            "link": "https://creativecommons.org/publicdomain/zero/1.0/",
            "icon": "📸"
        }
    }

    # 🧭 Hämta source-info
    if isinstance(source, str):
        src = sources.get(source.lower(), {"name": source, "type": "okänd", "domain": "", "copyright": "okänd", "link": "", "icon": "📍"})
    elif isinstance(source, dict):
        src = {**{"type": "okänd", "domain": "", "icon": "📍"}, **source}
    else:
        src = {"name": "Okänd källa", "type": "okänd", "domain": "", "copyright": "okänd", "link": "", "icon": "📍"}

    # Hjälpfunktion för domän
    def domain(u):
        try:
            return urlparse(u).netloc.lower()
        except Exception:
            return ""

    wd = domain(website)
    src_domain = src.get("domain", "")

    # ------------- Dynamiska länkar -------------
    osm_url = f"https://www.openstreetmap.org/{osm_type}/{osm_id}" if osm_id else ""
    mapillary = f"https://www.mapillary.com/app/?lat={lat:.5f}&lng={lon:.5f}&z=17" if lat and lon else ""
    google_map = f"https://www.google.com/maps?q={lat},{lon}" if lat and lon else ""
    google_photo = f"https://www.google.com/maps/@?api=1&map_action=pano&viewpoint={lat},{lon}" if lat and lon else ""
    commons = row.get("wikimedia_commons", "")
    commons_html = f"<a href='https://commons.wikimedia.org/wiki/{commons}' target='_blank'>📸 Wikimedia Commons</a><br>" if commons else ""

    # Bild
    img_html = f"<a href='{image}' target='_blank'><img src='{image}' width='220' style='border-radius:6px;'></a><br>" if image else ""

    # Webbplatslogik baserat på källa
    web_html = ""
    if website:
        if src["type"] == "kommun" and src_domain and wd.endswith(src_domain):
            web_html = f"<a href='{website}' target='_blank'>🌐 Officiell webbplats</a><br>"
        elif src["type"] == "crowdsourced" and src_domain in wd:
            web_html = f"<a href='{website}' target='_blank'>🐶 Hundlistan.se</a><br>"
        else:
            web_html = f"<a href='{website}' target='_blank'>🌐 Webbplats</a><br>"

    # Extra extern länk (t.ex. Hundlistan, om inte samma som website)
    extra_html = ""
    if url and url != website:
        if src["type"] == "crowdsourced" and "hundlistan.se" in url:
            extra_html = f"<a href='{url}' target='_blank'>🐶 Hundlistan.se</a><br>"
        else:
            extra_html = f"<a href='{url}' target='_blank'>🔗 Extern källa</a><br>"

    # Wikidata / Operator
    wiki_html = f"<a href='https://www.wikidata.org/wiki/{wikidata}' target='_blank'>🧭 Wikidata</a><br>" if wikidata else ""
    operator_html = f"<a href='https://www.wikidata.org/wiki/{operator_wd}' target='_blank'>🏢 Operator Wikidata</a><br>" if operator_wd else ""

    # Tillgänglighet
    wheelchair_html = ""
    if isinstance(tags, dict) and "wheelchair" in tags:
        val = str(tags["wheelchair"]).lower()
        wheelchair_html = {
            "yes": "♿ <b>Tillgänglig:</b> Ja ✅<br>",
            "limited": "♿ <b>Tillgänglig:</b> Begränsad ⚠️<br>",
            "no": "♿ <b>Tillgänglig:</b> Nej ❌<br>"
        }.get(val, "♿ <b>Tillgänglig:</b> Okänd ❔<br>")

    # OSM / Wheelmap
    osm_html = ""
    if osm_id:
        osm_label = f"{osm_type.capitalize()} {osm_id}"
        osm_html = (
            f"🆔 <b>OSM-ID:</b> <a href='{osm_url}' target='_blank'>{osm_label}</a><br>"
            f"<a href='https://wheelmap.org/{osm_type}/{osm_id}' target='_blank'>🗺️ Wheelmap</a><br>"
        )

    coords_html = (
        f"🧭 <b>Koordinater:</b> "
        f"<a href='https://www.openstreetmap.org/?mlat={lat:.5f}&mlon={lon:.5f}#map=18/{lat:.5f}/{lon:.5f}' target='_blank'>"
        f"{lat:.5f}, {lon:.5f}</a><br>"
        if (lat and lon) else ""
    )
    # 🆕 Källa & Copyright
    if src.get("link"):
        source_html = f"<hr><b>Källa:</b> {src['icon']} {src['name']}<br>📜 <b>Copyright:</b> <a href='{src['link']}' target='_blank'>{src['copyright']}</a>"
    else:
        source_html = f"<hr><b>Källa:</b> {src['icon']} {src['name']}<br>📜 <b>Copyright:</b> {src['copyright']}"

    # ------------- Slutlig popup -------------
    popup_html = f"""
    <div style='font-family:Arial; font-size:13px; max-width:300px;'>
      <b>{src['icon']} {name}</b><br>
      {desc}<br>
      {img_html}
      {web_html}{extra_html}
      <a href='{mapillary}' target='_blank'>🟢 Mapillary</a><br>
      <a href='{google_map}' target='_blank'>🗺️ Google Maps</a><br>
      <a href='{google_photo}' target='_blank'>📷 Street View</a><br>
      {wiki_html}{operator_html}{commons_html}
      {wheelchair_html}{osm_html}{coords_html}
      {source_html}
    </div>
    """
    return popup_html
```

```python
import requests, json, time, os
from bs4 import BeautifulSoup
OFFLINE_MODE = True

if OFFLINE_MODE:
    print("🚫 Offline-läge: inga nya data hämtas.")
    with open("hundparker_umap.geojson", "r", encoding="utf-8") as f:
        geojson = json.load(f)
    features = geojson["features"]
else:
    BASE_URL = "https://www.hundlistan.se/"
    OUTFILE = "hundparker_umap.geojson"
    
    params = {
        "mylisting-ajax": "1",
        "action": "get_listings",
        "security": "5e528da4d5",
        "form_data[preserve_page]": "true",
        "form_data[sort]": "latest",
        "form_data[region]": "",
        "form_data[search_location]": "",
        "form_data[lat]": "false",
        "form_data[lng]": "false",
        "listing_type": "hundpark",
        "listing_wrap": "col-md-12 grid-item"
    }
    
    # --- Om filen redan finns, ladda tidigare resultat (så du kan återuppta)
    if os.path.exists(OUTFILE):
        with open(OUTFILE, "r", encoding="utf-8") as f:
            geojson = json.load(f)
        features = geojson.get("features", [])
        print(f"🔁 Återupptar från tidigare körning ({len(features)} poster hittade).")
    else:
        features = []
    
    for page in range(1, 40):
        params["form_data[page]"] = str(page)
        print(f"Hämtar sida {page}/39 ...", end="\r")
    
        try:
            r = requests.get(BASE_URL, params=params, timeout=15)
            r.raise_for_status()
        except Exception as e:
            print(f"\n⚠️ Sida {page} misslyckades: {e}")
            continue
    
        data = r.json()
        soup = BeautifulSoup(data.get("html", ""), "html.parser")
    
        items = soup.select(".lf-item-container")
        if not items:
            print(f"\nℹ️ Inga fler resultat på sida {page}. Avslutar.")
            break
    
        for item in items:
            try:
                data_id = item.get("data-id")
                if any(f["properties"]["id"] == data_id for f in features):
                    continue  # hoppa över redan sparad post
    
                title = item.select_one(".listing-preview-title").get_text(strip=True)
                href = item.select_one("a")["href"]
                lan = item.select_one(".lf-contact li").get_text(strip=True)
                loc = json.loads(item.get("data-locations"))[0]
                address = loc.get("address")
                lat, lng = float(loc.get("lat")), float(loc.get("lng"))
    
                # --- Hämta detaljsida för "Hemsida"
                hemsida_url = None
                try:
                    detail = requests.get(href, timeout=10)
                    if detail.status_code == 200:
                        detail_soup = BeautifulSoup(detail.text, "html.parser")
                        hemsida_link = detail_soup.select_one("a[href*='http']:contains('Hemsida')")
                        if hemsida_link:
                            hemsida_url = hemsida_link["href"]
                except Exception as e:
                    print(f"⚠️ Kunde inte hämta detaljer för {title}: {e}")
    
   
                features.append({
                    "type": "Feature",
                    "geometry": {"type": "Point", "coordinates": [lng, lat]},
                    "properties": {
                        "id": data_id,
                        "namn": title,
                        "adress": address,
                        "lan": lan,
                        "url": href,
                        "hemsida": hemsida_url,
                    }
                })
    
            except Exception as e:
                print(f"\n⚠️ Fel vid parsing: {e}")
    
        # --- Spara efter varje sida (inkrementellt)
        geojson = {"type": "FeatureCollection", "features": features}
        with open(OUTFILE, "w", encoding="utf-8") as f:
            json.dump(geojson, f, ensure_ascii=False, indent=2)
        print(f"\n💾 Sparade {len(features)} poster efter sida {page}.")

        print(f"\n✅ Klart! Totalt {len(features)} hundparker sparade i {OUTFILE}")
```

## Hundlekplatsen 


```python
import json
import folium
from folium.plugins import BeautifyIcon

# 🗺️ Create (or reuse) map centered on Sweden
try:
    m
except NameError:
    m = folium.Map(location=[59.33, 18.06], zoom_start=6, tiles="CartoDB Positron")

# 🐕 FeatureGroup for Hundlekplatser
hundlekplatser_layer = folium.FeatureGroup(name="🐕 Hundlekplatser")

# 📂 Read GeoJSON file
with open("hundlekplatser_sverige.geojson", "r", encoding="utf-8") as f:
    data = json.load(f)

# 🔁 Iterate through each feature
for feature in data["features"]:
    props = feature["properties"]
    coords = feature["geometry"]["coordinates"]

    # 🧹 Prepare standardized row for popup
    row_dict = {
        "lat": coords[1],
        "lon": coords[0],
        "name": props.get("name", "Okänd hundlekplats"),
        "description": props.get("description", ""),
        "website": "",
        "url": props.get("url", ""),
        "image": props.get("image", ""),
        "wikidata": "",
        "operator:wikidata": "",
        "tags": {"city": props.get("city", "")},
        "type": "node",
        "id": "",
    }

    # ✅ Normalize and clean
    row_dict = safe_popup_row(row_dict)

    # 🧱 Build standardized popup
    popup_html = build_popup(row_dict,"hundlekplatsen")
    popup = folium.Popup(popup_html, max_width=300)

    # 🐾 Add marker with green paw icon
    folium.Marker(
        location=[coords[1], coords[0]],
        popup=popup,
        tooltip=props.get("name", ""),
        icon=BeautifyIcon(
            icon="paw",
            icon_shape="marker",
            border_color="#228B22",
            text_color="white",
            background_color="#32CD32",
            border_width=2,
            inner_icon_style="font-size:24px;"
        ),
    ).add_to(hundlekplatser_layer)

# ✅ Add layer to map
hundlekplatser_layer.add_to(m)

#display(m)
```

```python
import requests
import json
import os
import time

# 🗺️ Overpass API
overpass_url = "https://overpass-api.de/api/interpreter"
sweden_bbox = "55.0,10.5,69.2,24.2"

# 🔍 Fråga – alltid samma
overpass_query = f"""
[out:json][timeout:60];
nwr["leisure"="dog_park"]({sweden_bbox});
out center;
"""

# 💾 Cacheinställningar
cache_file = "osm_dogparks_cache.json"
#cache_ttl = 60 # 60 sekunder
cache_ttl = 60 * 60 * 24 * 7  # 7 dagar i sekunder


# 🧠 Funktion med cache
def get_osm_dogparks():
    if os.path.exists(cache_file):
        age = time.time() - os.path.getmtime(cache_file)
        if age < cache_ttl:
            print("📦 Läser hundrastgårdar från cache...")
            with open(cache_file, "r", encoding="utf-8") as f:
                return json.load(f)
        else:
            print("⚠️ Cache äldre än 7 dagar — hämtar ny data...")

    print("🌍 Hämtar hundrastgårdar från OSM...")
    response = requests.get(overpass_url, params={"data": overpass_query})
    response.raise_for_status()
    data = response.json()

    # 💾 Spara till cache
    with open(cache_file, "w", encoding="utf-8") as f:
        json.dump(data, f, ensure_ascii=False, indent=2)

    print(f"✅ Sparade {len(data['elements'])} objekt till cache.")
    return data


# 🚀 Kör
osm_data = get_osm_dogparks()
print(f"🐾 Hittade {len(osm_data['elements'])} hundrastgårdar.")
```

```python
# Steg 1: Samlar alla node-ID som används som byggstenar i en way/relation.
# Steg 2: Filtrerar bort dessa “interna” noder men behåller alla andra.
# Steg 3: Gör ett rent DataFrame med id, type, lat, lon och alla taggar.
import pandas as pd

# 1️⃣ Hämta alla node-ID som används i en way eller relation
used_node_ids = set()

for el in osm_data["elements"]:
    if el["type"] in ("way", "relation"):
        for ref in el.get("nodes", []):
            used_node_ids.add(ref)
        for member in el.get("members", []):
            if member.get("type") == "node":
                used_node_ids.add(member["ref"])

# 2️⃣ Behåll allt som inte är en node ELLER noder som inte används
filtered_elements = []
for el in osm_data["elements"]:
    if el["type"] != "node" or el["id"] not in used_node_ids:
        filtered_elements.append(el)

print(f"Före: {len(osm_data['elements'])}, Efter: {len(filtered_elements)}")

# 3️⃣ (valfritt) skapa DataFrame
rows = []
for el in filtered_elements:
    if el["type"] == "node":
        lat = el.get("lat")
        lon = el.get("lon")
    else:
        lat = el.get("center", {}).get("lat")
        lon = el.get("center", {}).get("lon")

    row = {
        "id": el.get("id"),
        "type": el.get("type"),
        "lat": lat,
        "lon": lon
    }
    row.update(el.get("tags", {}))
    rows.append(row)

dfOSM = pd.DataFrame(rows)
print(dfOSM.head())
```

```python
import requests
import time
import geojson
import pandas as pd
import json
import os

# ----------------------------------------------------------
# 1️⃣ Configuration
# ----------------------------------------------------------
ROOT_CATEGORY = "Dog_parks_in_Sweden"
HEADERS = {"User-Agent": "DogparkSwedenBot/1.2 (https://github.com/salgo60/Dogpark_Sweden)"}
API_URL = "https://commons.wikimedia.org/w/api.php"
SLEEP = 0.3  # polite delay
CACHE_DIR = "cache_commons"
RerunbWikiCommons = False  # <--- ändra till True om du vill hämta nytt från Commons

os.makedirs(CACHE_DIR, exist_ok=True)

# ----------------------------------------------------------
# 2️⃣ Helpers: Cache read/write
# ----------------------------------------------------------
def load_cache(filename):
    path = os.path.join(CACHE_DIR, filename)
    if os.path.exists(path):
        with open(path, "r", encoding="utf-8") as f:
            return json.load(f)
    return None

def save_cache(filename, data):
    path = os.path.join(CACHE_DIR, filename)
    with open(path, "w", encoding="utf-8") as f:
        json.dump(data, f, ensure_ascii=False, indent=2)

# ----------------------------------------------------------
# 3️⃣ Get subcategories recursively (cached)
# ----------------------------------------------------------
def get_subcategories(category, seen=None):
    """Recursively get all subcategories starting from a root."""
    cache_name = f"subcats_{category.replace(' ', '_')}.json"
    if not RerunbWikiCommons:
        cached = load_cache(cache_name)
        if cached:
            print(f"🗂️ Loaded cached subcategories for {category}")
            return cached

    if seen is None:
        seen = set()
    if category in seen:
        return []
    seen.add(category)

    print(f"📂 Scanning subcategories of: {category}")
    params = {
        "action": "query",
        "list": "categorymembers",
        "cmtitle": f"Category:{category}",
        "cmnamespace": 14,
        "cmlimit": "max",
        "format": "json"
    }

    subcats = []
    while True:
        r = requests.get(API_URL, params=params, headers=HEADERS, timeout=15)
        data = r.json()
        members = data.get("query", {}).get("categorymembers", [])
        for m in members:
            subname = m["title"].replace("Category:", "")
            subcats.append(subname)
            subcats += get_subcategories(subname, seen)

        if "continue" not in data:
            break
        params.update(data["continue"])
        time.sleep(SLEEP)

    subcats = list(set(subcats))
    save_cache(cache_name, subcats)
    return subcats

# ----------------------------------------------------------
# 4️⃣ Get all image files in a category (cached)
# ----------------------------------------------------------
def get_category_files(category):
    cache_name = f"files_{category.replace(' ', '_')}.json"
    if not RerunbWikiCommons:
        cached = load_cache(cache_name)
        if cached:
            print(f"🖼️ Loaded cached files for {category}")
            return cached

    params = {
        "action": "query",
        "list": "categorymembers",
        "cmtitle": f"Category:{category}",
        "cmnamespace": 6,
        "cmlimit": "max",
        "format": "json"
    }

    files = []
    while True:
        r = requests.get(API_URL, params=params, headers=HEADERS, timeout=15)
        data = r.json()
        files.extend(data.get("query", {}).get("categorymembers", []))
        if "continue" not in data:
            break
        params.update(data["continue"])
        time.sleep(SLEEP)

    save_cache(cache_name, files)
    return files

# ----------------------------------------------------------
# 5️⃣ Get coordinates for each file (cached)
# ----------------------------------------------------------
def get_file_coordinates(titles):
    cache_name = "file_coords.json"
    if not RerunbWikiCommons:
        cached = load_cache(cache_name)
        if cached:
            print("📍 Loaded cached coordinates.")
            return cached

    coords = []
    for i, title in enumerate(titles, 1):
        params = {
            "action": "query",
            "titles": title,
            "prop": "coordinates",
            "format": "json"
        }
        try:
            r = requests.get(API_URL, params=params, headers=HEADERS, timeout=10)
            data = r.json()
            for page in data.get("query", {}).get("pages", {}).values():
                coord = page.get("coordinates", [{}])[0]
                if "lat" in coord and "lon" in coord:
                    coords.append({
                        "title": title,
                        "lat": coord["lat"],
                        "lon": coord["lon"],
                        "pageid": page["pageid"],
                        "url": f"https://commons.wikimedia.org/wiki/{title.replace(' ', '_')}"
                    })
        except Exception as e:
            print(f"⚠️ Error fetching coords for {title}: {e}")
        if i % 20 == 0:
            print(f"📍 Processed {i}/{len(titles)} files...")
        time.sleep(SLEEP)

    save_cache(cache_name, coords)
    return coords

# ----------------------------------------------------------
# 6️⃣ Main pipeline
# ----------------------------------------------------------
print(f"🚀 Starting recursive fetch for {ROOT_CATEGORY}")
all_cats = [ROOT_CATEGORY] + get_subcategories(ROOT_CATEGORY)
print(f"📚 Found {len(all_cats)} total categories (including subcats)")

all_files = []
for cat in all_cats:
    files = get_category_files(cat)
    print(f"📸 {cat}: {len(files)} files")
    all_files.extend(files)

titles = list({f["title"] for f in all_files})
print(f"✅ Total unique files found: {len(titles)}")

coords = get_file_coordinates(titles)
print(f"🗺️ {len(coords)} geotagged images total")

# ----------------------------------------------------------
# 7️⃣ Save to CSV + GeoJSON
# ----------------------------------------------------------
dfCommons = pd.DataFrame(coords)
dfCommons.to_csv("dogparks_commons_images_recursive.csv", index=False)
print("💾 Saved dogparks_commons_images_recursive.csv")

features = [
    geojson.Feature(
        geometry=geojson.Point((row["lon"], row["lat"])),
        properties={"title": row["title"], "url": row["url"]}
    )
    for _, row in dfCommons.iterrows()
]

with open("dogparks_commons_images_recursive.geojson", "w") as f:
    geojson.dump(geojson.FeatureCollection(features), f, indent=2)
print("💾 Saved dogparks_commons_images_recursive.geojson")

print("🎉 Done!")
```

```python
import requests
import folium
import geopandas as gpd
import pandas as pd
from shapely.geometry import mapping
from folium.plugins import BeautifyIcon
from IPython.display import display

# 1️⃣ Hämta Södertäljes öppna data
url = "https://static-content.sodertalje.se/opendata/hundrastgardar.json"
response = requests.get(url)
response.raise_for_status()
dataS = response.json()

# 2️⃣ Konvertera till GeoDataFrame
gdfS = gpd.GeoDataFrame.from_features(dataS["features"], crs="EPSG:4326")
print("Kolumner:", gdfS.columns.tolist())
print("Antal objekt:", len(gdfS))

# 3️⃣ Korrekt centroid (projicera till SWEREF 99 TM)
gdfS_proj = gdfS.to_crs(epsg=3006)
centroids_proj = gdfS_proj.centroid
centroids = gpd.GeoSeries(centroids_proj, crs=3006).to_crs(epsg=4326)
gdfS["lat"] = centroids.y
gdfS["lon"] = centroids.x
print("✅ Centroids beräknade korrekt i EPSG:3006 → EPSG:4326")

# 4️⃣ Lägg till standardfält
gdfS["name"] = gdfS.get("Namn", "")
gdfS["website"] = gdfS.get("Webbsida", "")
gdfS["description"] = ""
gdfS["image"] = ""
gdfS["wikidata"] = ""
gdfS["type"] = "way"
gdfS["id"] = ""
gdfS["url"] = ""
gdfS["operator:wikidata"] = ""
gdfS["tags"] = [{} for _ in range(len(gdfS))]

# 5️⃣ Skapa lager
sodertalje_layer = folium.FeatureGroup(name="🐾 Södertälje kommun", show=True)

# 6️⃣ Iterera och skapa popup + markör
for _, row in gdfS.iterrows():
    geom = row.geometry
    if geom.is_empty:
        continue

    # ⚙️ Normalisera rad till dict
    row_dict = safe_popup_row(row)

    # 🧱 Skapa popup
    popup_html = build_popup(row_dict,"sodertalje")
    popup = folium.Popup(popup_html, max_width=300)

    # Polygon (yta)
    folium.GeoJson(
        data=mapping(geom),
        style_function=lambda x: {
            "fillColor": "#32CD32",
            "color": "#006400",
            "weight": 2,
            "fillOpacity": 0.4,
        },
        tooltip=row_dict.get("name", ""),
    ).add_to(sodertalje_layer)

    # Marker (centroid)
    folium.Marker(
        location=[row_dict["lat"], row_dict["lon"]],
        popup=popup,
        tooltip=row_dict.get("name", ""),
        icon=BeautifyIcon(
            icon="paw",
            icon_shape="marker",
            border_color="#006400",
            text_color="white",
            background_color="#32CD32",
            border_width=2,
            inner_icon_style="font-size:26px;",
        ),
    ).add_to(sodertalje_layer)

# 7️⃣ Lägg till lager & visa
sodertalje_layer.add_to(m)
# folium.LayerControl(collapsed=False).add_to(m)

#display(m)
```

### Borås kommun
* https://catalog.boras.se/store/1/resource/80


```python
import requests
import folium
import geopandas as gpd
import pandas as pd
from shapely.geometry import mapping
from folium.plugins import BeautifyIcon
from IPython.display import display

# 1️⃣ Hämta Borås öppna data
urlB = "https://catalog.boras.se/store/1/resource/80"
response = requests.get(urlB)
response.raise_for_status()
dataB = response.json()

# 2️⃣ Konvertera till GeoDataFrame
gdfB = gpd.GeoDataFrame.from_features(dataB["features"], crs="EPSG:4326")
print("Kolumner:", gdfB.columns.tolist())
print("Antal objekt:", len(gdfB))

# 3️⃣ Korrekt centroid via SWEREF 99 TM (EPSG:3006)
gdfB_proj = gdfB.to_crs(epsg=3006)
centroids_proj = gdfB_proj.centroid
centroids = gpd.GeoSeries(centroids_proj, crs=3006).to_crs(epsg=4326)
gdfB["lat"] = centroids.y
gdfB["lon"] = centroids.x
print("✅ Centroids korrekt beräknade (EPSG:3006 → EPSG:4326)")

# 4️⃣ Anpassa kolumner till popupformat
gdfB["name"] = gdfB.get("namn", "")
gdfB["website"] = gdfB.get("link", "")
gdfB["description"] = gdfB.get("Adress", "")
gdfB["image"] = ""
gdfB["wikidata"] = ""
gdfB["operator:wikidata"] = ""
gdfB["tags"] = [{} for _ in range(len(gdfB))]
gdfB["type"] = "way"
gdfB["id"] = gdfB.index.astype(str)
gdfB["url"] = gdfB.get("link", "")

# 5️⃣ Lager för Borås
boras_layer = folium.FeatureGroup(name="🐾 Borås kommun", show=True)

# 6️⃣ Lägg till polygoner + popupar
for _, row in gdfB.iterrows():
    geom = row.geometry
    if geom.is_empty:
        continue

    # Gör till dict för popup
    row_dict = safe_popup_row(row)

    # Popup HTML
    popup_html = build_popup(row_dict,"boras")
    popup = folium.Popup(popup_html, max_width=300)

    # Polygon
    folium.GeoJson(
        data=mapping(geom),
        style_function=lambda x: {
            "fillColor": "#C19ACB",
            "color": "#6B1E72",
            "weight": 2,
            "fillOpacity": 0.4,
        },
        tooltip=row_dict.get("name", ""),
    ).add_to(boras_layer)

    # Centroidmarkör 🐾
    folium.Marker(
        location=[row_dict["lat"], row_dict["lon"]],
        popup=popup,
        tooltip=row_dict.get("name", ""),
        icon=BeautifyIcon(
            icon="paw",
            icon_shape="marker",
            border_color="#6B1E72",
            text_color="white",
            background_color="#C19ACB",
            border_width=2,
            inner_icon_style="font-size:26px;",
        ),
    ).add_to(boras_layer)

# 7️⃣ Lägg till lager
boras_layer.add_to(m)
# folium.LayerControl(collapsed=False).add_to(m)

#display(m)
```

```python
import folium

def list_map_layers(m):
    print("Layers in map:")
    for key, val in m._children.items():
        layer_name = getattr(val, 'layer_name', '❌ No name')
        print(f"- {key:40} {layer_name:30} {val.__class__}")

list_map_layers(m)
```

```python
def layer_summary(m):
    print("\nLayer summary:")
    for key, val in m._children.items():
        if isinstance(val, folium.map.FeatureGroup):
            layer_name = getattr(val, 'layer_name', '❌ No name')
            count = sum(isinstance(c, folium.map.Marker) for c in val._children.values())
            print(f"{layer_name:35}  →  {count:3d} markers")

layer_summary(m)
```

```python
# -------------------------------------------------------------------
# 📊 Layer Summary (collapsible, bottom-left, skipping meta layers)
# -------------------------------------------------------------------
from collections import defaultdict

# Layers to skip from the summary
SKIP_LAYERS = [
    "🧩 Saknar data (OSM)",
    "🖼️ Wikimedia Commons-bilder (tagg + API)"
]

# Count markers/geometries in each FeatureGroup
layer_counts = defaultdict(int)
for key, layer in m._children.items():
    if isinstance(layer, folium.map.FeatureGroup):
        name = getattr(layer, "layer_name", "")
        if not name or name in SKIP_LAYERS:
            continue
        marker_count = len(layer._children)
        layer_counts[name] = marker_count

# Sort by descending count
sorted_layers = sorted(layer_counts.items(), key=lambda x: x[1], reverse=True)

# Build the summary HTML
summary_html = """
<style>
#layerSummary {
  position: fixed;
  bottom: 10px; left: 10px;
  z-index: 9998;
  background-color: rgba(255,255,255,0.96);
  padding: 10px 14px;
  border-radius: 12px;
  box-shadow: 0 0 8px rgba(0,0,0,0.3);
  font-family: Arial, sans-serif;
  font-size: 13px;
  max-width: 260px;
  transition: all 0.3s ease-in-out;
}
#summaryToggle {
  background-color: #0077cc;
  color: white;
  border: none;
  border-radius: 8px;
  padding: 5px 10px;
  cursor: pointer;
  font-size: 13px;
  width: 100%;
  text-align: center;
}
#summaryToggle:hover {
  background-color: #005fa3;
}
#summaryContent {
  display: none;
  margin-top: 8px;
  max-height: 280px;
  overflow-y: auto;
}
</style>

<div id="layerSummary">
  <button id="summaryToggle">📊 Visa lagerstatistik</button>
  <div id="summaryContent">
    <b>Layer summary:</b><br><ul style="padding-left:14px; margin-top:6px;">
"""

for name, count in sorted_layers:
    summary_html += f"<li>{name} → <b>{count}</b> objekt</li>"

summary_html += """
    </ul>
  </div>
</div>

<script>
  const btnS = document.getElementById('summaryToggle');
  const boxS = document.getElementById('summaryContent');
  let openS = false;
  btnS.onclick = () => {
    openS = !openS;
    boxS.style.display = openS ? 'block' : 'none';
    btnS.innerText = openS ? '❌ Dölj lagerstatistik' : '📊 Visa lagerstatistik';
  };
</script>
"""

m.get_root().html.add_child(folium.Element(summary_html))
```
