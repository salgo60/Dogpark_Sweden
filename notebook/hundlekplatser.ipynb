```python
import folium
import json

# Load the GeoJSON file
with open("hundlekplatser_sverige.geojson", "r", encoding="utf-8") as f:
    data = json.load(f)

# Center the map roughly on Sweden
m = folium.Map(location=[60.0, 15.0], zoom_start=5, tiles="OpenStreetMap")

# Loop through each feature
for feature in data["features"]:
    props = feature["properties"]
    coords = feature["geometry"]["coordinates"]

    name = props.get("name", "Unnamed")
    city = props.get("city", "")
    description = props.get("description", "")
    url = props.get("url", "")
    image = props.get("image", "")

    # Create popup HTML
    popup_html = f"""
    <div style="width:220px">
      <h4>{name}</h4>
      <p><b>City:</b> {city}</p>
      <p><b>Description:</b> {description}</p>
      <img src="{image}" width="200"><br>
      <a href="{url}" target="_blank">🔗 Open on hundlekplatsen.se</a>
    </div>
    """

    popup = folium.Popup(popup_html, max_width=250)
    folium.Marker(
        location=[coords[1], coords[0]],
        popup=popup,
        icon=folium.Icon(color="green", icon="paw", prefix="fa"),
    ).add_to(m)

# Display map in notebook
m.save("hundlekplatser_map.html")
m
```
